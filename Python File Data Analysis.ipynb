(Data Analysis) Question 2: Write a Python program to calculate and show correlation of all numeric variables. Display top 5 correlated feature.

```python
# Selecting only numeric columns for correlation calculation

numeric_features = traffic_clean.select_dtypes(include='number')

# Calculating the correlation matrix

correlation_matrix = numeric_features.corr()

print('Correlation Matrix (all numeric features):')
print()
print(correlation_matrix)
```

```python
# Finding the top 5 most correlated feature pairs

all_pairs = correlation_matrix.unstack()
all_pairs = all_pairs[all_pairs < 1] # removing self-correlation
all_pairs = all_pairs.abs().sort_values(ascending=False)
all_pairs = all_pairs.drop_duplicates()  

print('Top 5 Most Correlated Feature Pairs:')
print()
for (col1, col2), value in all_pairs.head(5).items():
    print(f'  {col1} & {col2} = {value}')
```

(Data Exploration) Question 1: Plot a bar chart showing the frequency of each category in the Label column (with values as Normal Traffic and Attack). Determine whether the Label, as the dependent variable, is balanced or not in the dataset. Provide an analysis.

```python
# Counting how many records belong to each label category

label_frequency = traffic_clean['Label'].value_counts()

# Plotting a bar chart

plt.figure(figsize=(7, 5))
plt.bar(label_frequency.index, label_frequency.values, color=['hotpink', 'mediumseagreen'], width=0.4)

# Adding the count value on top of each bar for clarity

for i, val in enumerate(label_frequency.values):
    plt.text(i, val + 1000, str(val), ha='center', fontsize=11)

plt.title('Frequency of Each Category in the Label Column', fontsize=13)
plt.xlabel('Label', fontsize=11)
plt.ylabel('Count', fontsize=11)
plt.tight_layout()
plt.show()
```

(Data Exploration) Question 2: Determine which category in the Label column (with values as Normal Traffic and Attack) has the highest average Flow Duration and the highest average Packet Length Mean. Visualize these findings with pie chart.

```python
# Calculating average Flow Duration and Packet Length Mean grouped by Label

avg_flow_duration = traffic_clean.groupby('Label')['Flow Duration'].mean()
avg_packet_length = traffic_clean.groupby('Label')['Packet Length Mean'].mean()

# Printing the values

print('Average Flow Duration by Label:')
print(avg_flow_duration.round(2))
print()
print('Average Packet Length Mean by Label:')
print(avg_packet_length.round(2))
print()

# Identifying which class has the highest value

print(f'Highest Average Flow Duration   : {avg_flow_duration.idxmax()}')
print(f'Highest Average Packet Length   : {avg_packet_length.idxmax()}')
```

```python
# Plotting two pie charts 

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Pie chart 1: Average Flow Duration

axes[0].pie(
    avg_flow_duration.values,
    labels=avg_flow_duration.index,
    autopct='%1.1f%%',
    colors=['hotpink', 'mediumseagreen'],
    startangle=90
)
axes[0].set_title('Average Flow Duration by Label', fontsize=12)

# Pie chart 2: Average Packet Length Mean

axes[1].pie(
    avg_packet_length.values,
    labels=avg_packet_length.index,
    autopct='%1.1f%%',
    colors=['hotpink', 'mediumseagreen'],
    startangle=90
)
axes[1].set_title('Average Packet Length Mean by Label', fontsize=12)

plt.tight_layout()
plt.show()
```

(Data Exploration) Question 4: Perform a hypothesis test to determine if there is a significant difference in the mean Flow Duration between Normal Traffic and Attack classes. State your null and alternative hypotheses, choose an appropriate test, perform it, and interpret the results.

```python
from scipy.stats import ttest_ind

# Separating Flow Duration values for each label class

normal_duration = traffic_clean[traffic_clean['Label'] == 'Normal Traffic']['Flow Duration']
attack_duration = traffic_clean[traffic_clean['Label'] == 'Attack']['Flow Duration']

# Printing the mean of each group

print(f'Mean Flow Duration - Normal Traffic : {normal_duration.mean()}')
print(f'Mean Flow Duration - Attack         : {attack_duration.mean()}')
print()

# Performing independent samples t-test

t_stat, p_value = ttest_ind(normal_duration, attack_duration, equal_var=False)

print(f'T-statistic : {t_stat}')
print(f'P-value     : {p_value}')
print()

# If p-value is less than 0.05 there is a significant difference between the two classes

if p_value < 0.05:
    print('Result: Reject H0.')
    print('There IS a statistically significant difference in mean Flow Duration between the two classes.')
else:
    print('Result: Fail to reject H0.')
    print('There is NO statistically significant difference in mean Flow Duration between the two classes.')
```
